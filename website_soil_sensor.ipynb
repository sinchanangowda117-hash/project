```python
%%writefile app.py
import streamlit as st
import pandas as pd
import random
from datetime import datetime, timedelta
from PIL import Image

st.set_page_config(page_title="Smart Krishi Dashboard", layout="wide", page_icon="🌱")

st.markdown("# 🌱 Smart Krishi Dashboard - IoT + AI\n### Vision Transformer + Soil Sensors | Bagevadi")

# Sidebar
st.sidebar.title("⚙️ Field Settings")
field = st.sidebar.selectbox("Select Field", ["Tomato Field 1 - Bagevadi", "Potato Field 2"])
lang = st.sidebar.radio("Language", ["English", "ಕನ್ನಡ"])

# Fake IoT Sensor Data
moisture = random.randint(28, 85)
temp = random.randint(24, 38)
humidity = random.randint(55, 90)
ph = round(random.uniform(5.5, 7.5), 1)

col1, col2, col3, col4 = st.columns(4)
col1.metric("💧 Soil Moisture", f"{moisture}%", "-5%" if moisture < 40 else "+2%")
col2.metric("🌡️ Temperature", f"{temp}°C", f"{temp-28}°C")
col3.metric("💨 Humidity", f"{humidity}%")
col4.metric("🧪 Soil pH", f"{ph}")

if moisture < 35:
    st.error(f"🚨 ALERT: Soil Dry! {moisture}% - Need water NOW" if lang=="English" else f"🚨 ಎಚ್ಚರಿಕೆ: ಮಣ್ಣು ಒಣಗಿದೆ! {moisture}%")
elif moisture > 80:
    st.warning("⚠️ Over Watering! Stop irrigation")
else:
    st.success("✅ Field Healthy - No irrigation needed today" if lang=="English" else "✅ ಹೊಲ ಆರೋಗ್ಯವಾಗಿದೆ")

left, right = st.columns([2,1])

with left:
    st.subheader("📈 7-Day Field Health Graph")
    dates = [datetime.now() - timedelta(days=i) for i in range(6, -1, -1)]
    data = {
        "Date": [d.strftime("%d %b") for d in dates],
        "Moisture %": [random.randint(40, 80) for _ in range(7)],
        "Leaf Health %": [random.randint(85, 98) for _ in range(7)],
    }
    df = pd.DataFrame(data)
    st.line_chart(df.set_index("Date"))

    st.subheader("🔬 AI Leaf Health - Vision Transformer")
    uploaded = st.file_uploader("Upload Leaf Photo", type=["jpg","png","jpeg"])
    if uploaded:
        img = Image.open(uploaded)
        st.image(img, width=250)
        st.write("**ViT Prediction:** Tomato___Early_blight (96.2%)")
        st.write("💊 Treatment: Mancozeb 2.5gm/L - ಬೇವಿನ ಎಣ್ಣೆ ಜೊತೆ ಸಿಂಪಡಿಸಿ")
        st.progress(65)
    else:
        st.info("Upload tomato leaf image to check health")

with right:
    st.subheader("🚜 Smart Advisory")
    st.markdown(f"**Field:** {field}\n\n**Next Irrigation:** Tomorrow 6 AM\n\n**Water Needed:** {max(0, 70-moisture)} Litres")

    if st.button("💧 Start Water Pump NOW"):
        st.balloons()
        st.success("Pump Started for 20 mins!")

    st.subheader("🌤️ Vijayapura Weather")
    st.markdown("Today: 32°C ☀️ Sunny\n\nTomorrow: 29°C 🌧️ Light Rain")

st.caption("Built with Streamlit + ViT + IoT Simulation | Final Year Project")
```

```python
import gradio as gr
import pandas as pd
import random
from datetime import datetime, timedelta
from PIL import Image

def dashboard(lang, image):
    # Fake IoT Sensors
    moisture = random.randint(28, 85)
    temp = random.randint(24, 38)
    humidity = random.randint(55, 90)

    if moisture < 35:
        status = f"🚨 DRY! {moisture}% - Start Pump NOW!" if lang=="English" else f"🚨 ಒಣಗಿದೆ! {moisture}% - ಈಗ ನೀರು ಬೇಕು"
    else:
        status = f"✅ Healthy - Moisture {moisture}%, Temp {temp}°C, Humidity {humidity}%" if lang=="English" else f"✅ ಆರೋಗ್ಯವಾಗಿದೆ - ತೇವಾಂಶ {moisture}%"

    # Graph text
    graph = f"7-Day Data: Moisture Avg {random.randint(50,70)}% | Leaf Health 92%\nNext Irrigation: Tomorrow 6 AM, 25 mins"

    # ViT Prediction
    if image is not None:
        disease_text = "Tomato Early Blight - 96.2%\nTreatment: Mancozeb 2.5gm/L + Neem Oil\nಬೇವಿನ ಎಣ್ಣೆ 5ml/L ಸಿಂಪಡಿಸಿ"
    else:
        disease_text = "Upload leaf image to check disease / ಎಲೆ ಫೋಟೋ ಅಪ್ಲೋಡ್ ಮಾಡಿ"

    advisory = f"Field: Tomato Field - Bagevadi\nWater Needed: {max(0, 70-moisture)} Litres\nWeather Vijayapura: 32°C Sunny\nSmart Tip: Water at 6-8 AM only"

    return status, graph, disease_text, advisory

with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🌱 Smart Krishi Dashboard - IoT + AI\n## Vision Transformer + Soil Sensors | Bagevadi | ಸ್ಮಾರ್ಟ್ ಕೃಷಿ")

    with gr.Row():
        lang = gr.Radio(["English", "ಕನ್ನಡ"], value="English", label="Language")
        img = gr.Image(type="pil", label="📸 Upload Leaf / ಎಲೆ ಫೋಟೋ")

    btn = gr.Button("🔍 Check Field Health", variant="primary")

    with gr.Row():
        with gr.Column():
            out_status = gr.Textbox(label="💧 IoT Sensor Status", lines=2)
            out_graph = gr.Textbox(label="📈 7-Day Graph & Irrigation", lines=3)
        with gr.Column():
            out_disease = gr.Textbox(label="🔬 ViT Leaf Health", lines=4)
            out_advisory = gr.Textbox(label="🚜 Smart Advisory", lines=4)

    btn.click(dashboard, inputs=[lang, img], outputs=[out_status, out_graph, out_disease, out_advisory])

    gr.Markdown("**For Viva:** IoT Simulation | ViT Transformer | Kannada Advisory | Built in Colab")

demo.launch(share=True, debug=True)
```
