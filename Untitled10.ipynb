```python
from langchain_community.document_loaders import WebBaseLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS

# 1. Scrape Live News (Change links to any news you want)
news_urls = [
    "https://www.bbc.com/news/technology",
    "https://timesofindia.indiatimes.com/topic/artificial-intelligence",
]

print("Scraping today's live news...")
loader = WebBaseLoader(news_urls)
docs = loader.load()

# 2. Split into chunks
splitter = RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=100)
split_docs = splitter.split_documents(docs)

# 3. Build RAG DB (Free model, No API Key)
print(f"Found {len(split_docs)} news chunks. Building RAG...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
db = FAISS.from_documents(split_docs, embeddings)

print(f"✅ NewsRAG Ready with TODAY'S News!")
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# 1. Scrape Live News
news_urls = [
    "https://www.bbc.com/news/technology",
]

print("Scraping today's live news...")
loader = WebBaseLoader(news_urls)
docs = loader.load()

# 2. Split
splitter = RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=100)
split_docs = splitter.split_documents(docs)

# 3. Build RAG DB
print(f"Found {len(split_docs)} chunks. Building RAG...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
db = FAISS.from_documents(split_docs, embeddings)

print(f"✅ NewsRAG Ready with TODAY'S News!")
```

```python
import gradio as gr

def ask_news(question):
    results = db.similarity_search(question, k=3)
    context = "\n\n".join([r.page_content[:600] for r in results])

    answer = f"📰 Retrieved from TODAY'S Live News:\n\n{context}\n\n---\n✅ Answer for '{question}': Based on above live news, this is the latest info."
    return answer

demo = gr.Interface(
    fn=ask_news,
    inputs=gr.Textbox(label="Ask about TODAY'S News", placeholder="e.g., What happened in AI today?"),
    outputs=gr.Textbox(label="Answer from Live News RAG", lines=15),
    title="NewsRAG - Live Daily News Chatbot",
    description="Answers ONLY from today's BBC news, not old ChatGPT memory."
)

demo.launch()
```
